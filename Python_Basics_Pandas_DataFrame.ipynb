```python
num = 7

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

**Example List**

**Question 9 : Create a random array of 10 numbers and:**

 - **Find max and min**

 - **Sort the array**

```python
import numpy as np

arr = np.random.randint(1, 100, 10)

print("Random Array:", arr)

print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))

print("Sorted Array:", np.sort(arr))
```

**Question 10 : Replace all even numbers with 0:**

**arr = np.array([1,2,3,4,5,6])**

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

arr[arr % 2 == 0] = 0

print(arr)
```
